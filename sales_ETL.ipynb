```python
import pandas as pd

from sqlalchemy import create_engine
```

```python
sales_df = pd.read_csv('/content/Retail and wherehouse Sale.csv')
```

```python
sales_df.isnull().sum()
```

```python
sales_df['ID'] = range(1, len(sales_df) + 1)
```

```python
sales_df.columns = sales_df.columns.str.replace(' ', '_')
```

```python
sales_df.head()
```

```python
# 1. Combine YEAR and MONTH into a new column 'SALE_DATE'

sales_df['SALE_DATE'] = pd.to_datetime(sales_df['YEAR'].astype(str) + '-' + sales_df['MONTH'].astype(str) + '-01')

# 2. Clean SUPPLIER and ITEM TYPE (make them uppercase and strip spaces)
sales_df['SUPPLIER'] = sales_df['SUPPLIER'].str.strip().str.upper()
sales_df['ITEM_TYPE'] = sales_df['ITEM_TYPE'].str.strip().str.upper()

# 3. Clean ITEM DESCRIPTION (just strip spaces)
sales_df['ITEM_DESCRIPTION'] = sales_df['ITEM_DESCRIPTION'].str.strip()

# 4. Fill missing values in sales columns with 0
sales_cols = ['RETAIL_SALES', 'RETAIL_TRANSFERS', 'WAREHOUSE_SALES']
sales_df[sales_cols] = sales_df[sales_cols].fillna(0)

# 5. Ensure sales columns are numbers (just in case)
sales_df[sales_cols] = sales_df[sales_cols].astype(float)

# 6. (Optional) Reorder columns if you want
sales_df = sales_df[['ID', 'YEAR', 'MONTH', 'SUPPLIER', 'ITEM_CODE', 'ITEM_DESCRIPTION', 'ITEM_TYPE',
         'RETAIL_SALES', 'RETAIL_TRANSFERS', 'WAREHOUSE_SALES']]

# Show the transformed data
print(sales_df.head())
```

```python
sales_df.to_csv('final_data.csv', index=False)
```
